```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("uscis_i140.csv")

# Merge category and subcategory for plotting clarity
df['Label'] = df['Category'] + " - " + df['Subcategory']
df['Time'] = df['Year'].astype(str) + " " + df['Quarter']

# Plot approvals
plt.figure(figsize=(12, 6))
sns.lineplot(data=df, x='Time', y='Received', hue='Label', marker='o')
plt.title("I-140 Approvals by Category and Quarter (2025)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
